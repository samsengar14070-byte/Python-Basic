# If else

```python
number = 7

if number % 2 == 0:
    print("The number is even.")
else:
    print("The number is odd.")
```

# If-elif-else

```python
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: F")
```

# Combining Conditions (Using Logical Operators)

```python
age = 25
experience = 5

# Both conditions must be True
if age >= 21 and experience >= 2:
    print("Eligible for the senior role.")
else:
    print("Does not meet the requirements.")
```
