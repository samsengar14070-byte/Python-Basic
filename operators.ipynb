# Identity 

```python
# Checks if variables reference the same memory location
# ==========================================
print("\n--- Identity Operators ---")
list_a = [1, 2, 3]
list_b = [1, 2, 3]
list_c = list_a

# 'is' checks if they are the exact same object in memory
# '==' checks if their contents are equal
print(f"list_a == list_b: {list_a == list_b}")  # True (same values)
print(f"list_a is list_b: {list_a is list_b}")  # False (different memory addresses)
print(f"list_a is list_c: {list_a is list_c}")  # True (points to the same object)

# 'is not' checks for non-identity
print(f"list_a is not list_b: {list_a is not list_b}") # True
```

# Membership

```python
print("\n--- Membership Operators ---")
fruits = ["apple", "banana", "cherry"]

# 'in' operator
has_apple = "apple" in fruits
print(f"Is 'apple' in fruits? {has_apple}")  # True

# 'not in' operator
has_grape = "grape" not in fruits
print(f"Is 'grape' missing from fruits? {has_grape}")  # True
```

# Bitwise

```python
# 0101 ^ 0011 = 0110 (Decimal: 6)
print(f"a ^ b  = {a ^ b}")

# Bitwise NOT (~): Inverts all the bits
# Formula: ~x = -(x + 1)
print(f"~a     = {~a}")

# Bitwise Left Shift (<<): Shifts bits to the left, filling with 0
# 0101 << 1 = 1010 (Decimal: 10)
print(f"a << 1 = {a << 1}")

# Bitwise Right Shift (>>): Shifts bits to the right
# 0101 >> 1 = 0010 (Decimal: 2)
print(f"a >> 1 = {a >> 1}")
```
