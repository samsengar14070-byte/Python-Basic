```python
# Empty list
empty_list = []

# List of integers
numbers = [1, 2, 3, 4, 5]

# List with mixed data types
mixed_list = ["Python", 3.14, True, 42]

# List of lists (Nested list)
matrix = [[1, 2], [3, 4]]

# Using the list() constructor
chars = list("Hello")  # Result: ['H', 'e', 'l', 'l', 'o']
```

 Accessing Items

```python
fruits = ["apple", "banana", "cherry", "date"]

print(fruits[0])   # Output: "apple" (First item)
print(fruits[-1])  # Output: "date" (Last item)

# Slicing (extracting a sublist)
print(fruits[1:3]) # Output: ['banana', 'cherry']
```

Adding item

```python
items = ["apple", "banana"]

# append(): Adds an item to the end
items.append("cherry") 
# items is now ['apple', 'banana', 'cherry']

# insert(): Adds an item at a specific index
items.insert(1, "blueberry") 
# items is now ['apple', 'blueberry', 'banana', 'cherry']

# extend(): Merges another iterable into the list
items.extend(["date", "elderberry"]) 
# items is now ['apple', 'blueberry', 'banana', 'cherry', 'date', 'elderberry']
```

Remove item

```python
items = ["apple", "banana", "cherry", "banana"]

# remove(): Deletes the first occurrence of a value
items.remove("banana") 
# items is now ['apple', 'cherry', 'banana']

# pop(): Removes and returns the item at a specific index (default is the last item)
popped_item = items.pop(1) 
# popped_item is 'cherry', items is now ['apple', 'banana']

# clear(): Empties the list completely
items.clear() 
# items is now []
```

Utility Methods & Operations

```python
numbers = [4, 2, 9, 2, 1]

# len(): Find the length of the list
print(len(numbers))  # Output: 5

# count(): Count occurrences of a specific item
print(numbers.count(2))  # Output: 2

# check if an item exists
if 9 in numbers:
    print("Found 9!")

# sort(): Sorts the list in place
numbers.sort() 
# numbers is now [1, 2, 2, 4, 9]

# reverse(): Reverses the elements in place
numbers.reverse() 
# numbers is now [9, 4, 2, 2, 1]
```

List Comprehension

```python
# Generate a list of squares for numbers 0 to 4
squares = [x**2 for x in range(5)]
# Output: [0, 1, 4, 9, 16]

# Filter a list (only keep even numbers)
evens = [x for x in range(10) if x % 2 == 0]
# Output: [0, 2, 4, 6, 8]
```
