# Dictionary

```python
prices ={"apple" : 1.20, "banana": 0.50, "cherry":2.50}

prices["banana"] = 0.65
print(prices)
prices["date"]=3.00
print(prices)
total_cost = 0
for price in prices.values():
    total_cost +=price
    
print("total cost:",(total_cost))
```
