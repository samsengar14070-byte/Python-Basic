# for loop

```python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
```

# Range function

```python
# Prints 0, 1, 2, 3, 4
for i in range(5):
    print(i)

# Count from 1 to 10 by twos (Prints 1, 3, 5, 7, 9)
for x in range(1, 11, 2):
    print(x)
```

# while loop

```python
count = 0

while count < 3:
    print(f"Count is {count}")
    count += 1  # Crucial to update the condition, otherwise it runs forever!
```

#  Loop Control Statements

```python
# Example of break
for num in range(10):
    if num == 5:
        break  # Stops the loop completely when num hits 5
    print(num)

# Example of continue
for num in range(5):
    if num == 2:
        continue  # Skips printing 2 and goes straight to 3
    print(num)
```

# Nested loop

```python
for row in range(2):
    for col in range(3):
        print(f"Row {row}, Col {col}")
```
